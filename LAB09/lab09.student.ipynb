```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_val_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
from sklearn import preprocessing
import matplotlib.pyplot as plt
```

```python
train = pd.read_csv('train.csv')
val = pd.read_csv('validation.csv')
```

```python
x_train = np.asarray(train)
x_val = np.asarray(val)
# x_train = np.asarray(train.iloc[:,5]).reshape(-1,1)
# x_val = np.asarray(val.iloc[:,5]).reshape(-1,1)
x_train = x_train.sum(axis=1).reshape(-1,1) / 100
x_val = x_val.sum(axis=1).reshape(-1,1) / 100
```

```python
x_train
```

```python
train['y']=np.zeros_like(train.grade)
val['y']=np.zeros_like(val.grade)
grades = ['F','D','D+','C','C+','B','B+','A']
grade_indexes = {grade: i for i, grade in enumerate(grades)}
# train.grade['F'] map to grade_indexes['F']
train.y = train.grade.map(lambda x:grade_indexes[x])
# for i in range(len(grades)):
#     train.y[train.grade == grades[i]] = i 

val.y = val.grade.map(lambda x:grade_indexes[x])

y_train = np.asarray(train.y)
y_val = np.asarray(val.y)
```

```python
np.random.seed(3)
# model = MLPClassifier(activation='relu', solver='sgd',
# hidden_layer_sizes=(5,5), max_iter=10000, early_stopping=True) 
model = MLPClassifier(activation='identity', solver='lbfgs',
hidden_layer_sizes=(2,2), max_iter=10000, early_stopping=True)
model.fit(x_train, y_train)
y_pred = model.predict(x_val)
accuracy = accuracy_score(y_val, y_pred)
print(f'Accuracy: {accuracy*100:.2f}%')
print(y_pred)
# print(model.loss_curve_)

# plt.plot(model.loss_curve_)
```
