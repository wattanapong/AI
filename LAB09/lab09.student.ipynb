```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
from sklearn import preprocessing
import matplotlib.pyplot as plt
```

```python
train = pd.read_csv('StudentScore.Train.csv' )
test = pd.read_csv('StudentScore.Test.csv')
```

```python
x_train = np.asarray(train.iloc[:,:6])
x_test = np.asarray(test.iloc[:,:6])
# x_train = np.asarray(train.iloc[:,5]).reshape(-1,1)
# x_test = np.asarray(test.iloc[:,5]).reshape(-1,1)
x_train = x_train.sum(axis=1).reshape(-1,1) / 100
x_test = x_test.sum(axis=1).reshape(-1,1) / 100
```

```python
train['y']=np.zeros_like(train.grade)
test['y']=np.zeros_like(test.grade)
grades = ['F','D','D+','C','C+','B','B+','A']
grade_indexes = {grade: i for i, grade in enumerate(grades)}
# train.grade['F'] map to grade_indexes['F']
train.y = train.grade.map(lambda x:grade_indexes[x])
# for i in range(len(grades)):
#     train.y[train.grade == grades[i]] = i 

test.y = test.grade.map(lambda x:grade_indexes[x])

y_train = np.asarray(train.y)
y_test = np.asarray(test.y)
```

```python
np.random.seed(3)
# model = MLPClassifier(activation='relu', solver='sgd',
# hidden_layer_sizes=(5,5), max_iter=10000, early_stopping=True) 
model = MLPClassifier(activation='identity', solver='lbfgs',
hidden_layer_sizes=(2,2), max_iter=10000, early_stopping=True)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy*100:.2f}%')
print(y_pred)
# print(model.loss_curve_)

# plt.plot(model.loss_curve_)
```
